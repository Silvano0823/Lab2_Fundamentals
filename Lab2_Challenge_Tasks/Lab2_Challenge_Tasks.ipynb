```python

# Exercise 1: Sensor Readings

SURNAME = "Silvano"
SEED_NUM = 0

print("Surname:", SURNAME)
print("Seed Number:", SEED_NUM)
```

```python

# Generate Sensor Readings

sensor_readings = [
    len(SURNAME) + SEED_NUM,
    len(SURNAME) * 2 + SEED_NUM,
    len(SURNAME) * 3 + SEED_NUM,
    len(SURNAME) * 4 + SEED_NUM,
    len(SURNAME) * 5 + SEED_NUM
]

print("Generated Sensor Data:", sensor_readings)
```

```python

# Validate Sensor Readings

valid_readings = []
invalid_readings = []

for reading in sensor_readings:
    if isinstance(reading, (int, float)) and reading >= 0:
        valid_readings.append(reading)
    else:
        invalid_readings.append(reading)

print("Valid Readings:", valid_readings)
print("Invalid Readings:", invalid_readings)
```

```python

# Classify Sensor Readings

limit = (SEED_NUM * 10) + 20

print("Safety Limit:", limit)

for reading in valid_readings:
    if reading > limit:
        print(reading, "- ALERT: Over Limit")
    elif reading < 5:
        print(reading, "- ALERT: Low Signal")
    else:
        print(reading, "- STATUS: Nominal")
```

```python

# Summary of Sensor Readings

print("===== SENSOR READINGS SUMMARY =====")
print("Surname:", SURNAME)
print("Seed Number:", SEED_NUM)
print("Generated Sensor Data:", sensor_readings)
print("Number of Valid Readings:", len(valid_readings))
print("Number of Invalid Readings:", len(invalid_readings))
print("Safety Limit:", limit)
print("Exercise 1 Completed!")
```

```python

# Exercise 2: Signal Analysis

# Generate a unique signal
signal = SURNAME + str(SEED_NUM) + "_SIGNAL"

print("Generated Signal:", signal)
```

```python

# Normalize the Signal

processed_signal = signal.upper()

print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)
```

```python

# Analyze Each Character

print("===== CHARACTER ANALYSIS =====")

for character in processed_signal:
    if character.isalpha():
        print(character, "- Letter")
    elif character.isdigit():
        print(character, "- Digit")
    else:
        print(character, "- Special Character")
```

```python

# Classify the Signal

letter_count = 0
digit_count = 0
special_count = 0

for character in processed_signal:
    if character.isalpha():
        letter_count += 1
    elif character.isdigit():
        digit_count += 1
    else:
        special_count += 1

if special_count > 0:
    signal_classification = "Alphanumeric Signal with Special Characters"
elif digit_count > 0:
    signal_classification = "Alphanumeric Signal"
else:
    signal_classification = "Alphabetic Signal"

print("===== SIGNAL CLASSIFICATION =====")
print("Letters:", letter_count)
print("Digits:", digit_count)
print("Special Characters:", special_count)
print("Signal Classification:", signal_classification)
```

```python

# Final Diagnostic Report

print("===== SIGNAL ANALYSIS REPORT =====")
print("Surname:", SURNAME)
print("Seed Number:", SEED_NUM)
print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)
print("Letters:", letter_count)
print("Digits:", digit_count)
print("Special Characters:", special_count)
print("Signal Classification:", signal_classification)
print("Status: Analysis Completed")
```

```python

# Exercise 3: Password Authentication

generated_password = SURNAME + str(SEED_NUM) + "!TUP"

attempt_limit = SEED_NUM + 3

print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
```

```python

# Manage Login Attempts

attempts_made = 0
access_granted = False

while attempts_made < attempt_limit:
    entered_password = input("Enter Password: ")
    attempts_made += 1

    if entered_password == generated_password:
        access_granted = True
        print("Access Granted!")
        break
    else:
        print("Incorrect Password!")

        remaining = attempt_limit - attempts_made
        print("Attempts Remaining:", remaining)

print("Attempts Made:", attempts_made)
```

```python

# Determine Final System State

if access_granted:
    system_state = "UNLOCKED"
else:
    system_state = "LOCKED"

print("===== AUTHENTICATION SUMMARY =====")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)

if access_granted:
    print("Access Result: Granted")
else:
    print("Access Result: Denied")

print("Final System State:", system_state)
```
